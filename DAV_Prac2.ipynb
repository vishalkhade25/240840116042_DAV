**Practical 2 | Perform Data inspection, cleaning, handling, missing values, duplicate removal and data transformation using pandas**

**Step 1 : Import Pandas**

```python
import pandas as pd
```

```python
data = {
    'ID' : [101, 102, 103, 104, 104, 105,106, 107],
    "Name" : ['Rahul', 'Priya', 'Amit',None, 'Riya', 'Ajay', 'Naresh', 'Jenil'],
    "Age" : [20,21,None,20,20,22,23,28],
    "City" : ['Surat','Navsari','Surat','Bharuch','Bharuch','Navsari', 'Surat', 'Ahmedabad'],
    "Marks" : [85, None, 78,88,88,81,90,93]
}
```

```python
df = pd.DataFrame(data)
df
```

**Step 3 : Inspect the Dataset**

```python
df.head()
```

**Step 4 : Check Missing Values**

```python
# Check for missing (null) values in every column
df.isnull()
# df.isnull().sum()
```

```python
# Count null values of each column
df.isnull().sum()
```

```python
#Total null values in whole dataframe

df.isnull().sum().sum()
```

**Step 5 : Fill Missing Values**

```python
#Replace missing values in the Marks column.
# mean() calculate the average marks.
# fillna() replaces missing values with the calculated average
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# display the updated Dataframe
df
```

**Fill Missing Age**

```python
# Replace missing values in Age column
# using the average age of all students
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

**Fill Missing Name**

```python
# Replace the missing names with "Unknown"
df["Name"] = df["Name"].fillna("Unknown")
df
```

**Step 6 : Detect duplicate rows**

```python
# Check whether each row is duplicated
df.duplicated()
```

```python
# Count duplicate row
df.duplicated().sum()
```

**Step 7 : Remove Duplicate Records**

```python
df = df.drop_duplicates()
df
```

**Step 8 : Rename a column**

```python
# Rename the column "Marks" to "Score"
# inplace=True permanently updates the DataFrame

df.rename(columns={"Marks":"Score"}, inplace=True)

# Display the updated DataFrame
df
```

**Step 9 : Change Data Type**

```python
# Convert the Age column from float to integer
# This removes the decimal part after filling missing values

df['Score'] = df['Score'].astype(int)
df.dtypes
```

```python
# Create a new column
# Add 5 marks to the column "Score"
df['Final_Score'] = df['Score'] + 5
df
```

```python
# Conversion of name from given case to uppercase
df['Name'] = df['Name'].str.upper()
df
```

```python
# Conversion of city from given case to lowercase
df['City'] = df['City'].str.lower()
df
```

```python
# Removing extra space
df['City'] = df['City'].str.strip()
df
```

```python
# sorting the data based on "Final Score"
df.sort_values(by="Final_Score", ascending=False)
```

```python
# Save the clean dataset
df.to_csv("clean_student_data.csv", index=False)
print("Dataset saved successfully")
```
